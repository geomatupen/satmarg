```python
from satmarg.core import get_precise_overpasses

# Location: example is Kathmandu
df = get_precise_overpasses(
    lat= 47.899167,
    lon= 17.007472,
    # lat = 27.700769,
    # lon = 85.300140,
    start_date="2025-12-22",
    end_date="2026-01-22",
    timezone="Europe/Vienna",
    satellites="SENTINEL-3A, SENTINEL-3B, LANDSAT 8, LANDSAT 9, ISS (ZARYA)",
    step_seconds=1,
    # max_angle_deg=0.6,
    max_angle_deg=None,
    output_format='csv' # options: csv, table, json
)

print(df)
```

```python
from datetime import datetime
import os

today = datetime.today().date();
print(today)
output_dir = 'Outputs'
os.makedirs(output_dir, exist_ok=True)
df.to_csv(f"Outputs/prediction_{today}.csv", index=False)
print("Data added to CSV")
```
