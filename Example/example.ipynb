```python
from satmarg.core import get_precise_overpasses

# Location: example is Kathmandu
df = get_precise_overpasses(
        # lat= 47.899167,
        # lon= 17.007472,
        lat = 27.700769,
        lon = 85.300140,
        start_date="2026-01-01",
        end_date="2026-02-01",
        timezone="Asia/Kathmandu",
        satellites = "SENTINEL-2B",
        # satellites = "SENTINEL-2B, SENTINEL-2C, SENTINEL-3A, SENTINEL-3B, LANDSAT 8, LANDSAT 9, ISS (ZARYA)", #single or multiple
        max_angle_deg = "0.7", # or "0.7, 0.7, 0.5, 0.5, 0.7, 0.7, 0.5",  #single (same for all) or multiple (count should match with no. of satellites)
        step_seconds=10,
        output_format='json',    
    )

print(df)
```

```python
from datetime import datetime
import os

today = datetime.today().date();
print(today)
output_dir = 'Outputs'
os.makedirs(output_dir, exist_ok=True)
df.to_csv(f"Outputs/prediction_{today}.csv", index=False)
print("Data added to CSV")
```
